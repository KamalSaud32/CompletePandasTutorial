# What is the definition of pandas?

Pandas is an open-source Python library designed for data manipulation and analysis. It provides powerful, easy-to-use data structures such as Series and DataFrame that allow users to work efficiently with structured data like tables, spreadsheets, and time-series data. Pandas supports reading and writing data from multiple file formats, including CSV, Excel, JSON, and databases. It offers tools for cleaning, transforming, filtering, aggregating, and analyzing data with minimal code. Built on top of NumPy, pandas is widely used in data science, machine learning, finance, and research because it simplifies handling real-world, messy datasets.

# Data Frame

A DataFrame is a two-dimensional data structure in pandas that organizes data into rows and columns, similar to a spreadsheet or database table. It allows efficient storage, manipulation, filtering, and analysis of structured data, making it easy to work with real-world datasets in Pytho

# Parquet in dataset:

Parquet is a columnar storage file format designed to efficiently store large datasets. Unlike row-based formats like CSV, Parquet stores data column by column, which allows for:

- Faster data reading for specific columns

- Better compression and reduced storage size

- Preservation of data types (integers, floats, strings, dates, etc.)

- It is widely used in data engineering, analytics, and big data platforms like Spark, Hive, and pandas.

| Parameter             | Meaning                          |
| --------------------- | -------------------------------- |
| `left`, `right`       | DataFrames to merge              |
| `how`                 | Type of join                     |
| `on`                  | Column(s) to join on             |
| `left_on`, `right_on` | Different column names           |
| `suffixes`            | Resolve overlapping column names |


**Inner join**

- Keeps only matching keys

**Left Jion**

✔ All rows from left
✔ Matching rows from right
✔ Missing matches → NaN

**Right join**

✔ All rows from right

**Outer join**

✔ Keeps all rows
✔ Missing matches → NaN
